# Open News demo for Google Colab

This notebook installs the project, fetches a few stories, searches by keyword, and exports a sample result.


```python
!pip install -q git+https://github.com/alphap365/open-news.git

from open_news import fetch, search, export

articles = fetch(category='tech', max_results=5)
print('Fetched articles:', len(articles))
for item in articles[:3]:
    print(item.get('title'), '->', item.get('url'))

results = search('AI', country='in', max_results=5)
print('Search results:', len(results))
for item in results[:3]:
    print(item.get('title'))
```

```python
from open_news import fetch, rank_articles, cluster_articles

items = fetch(category='tech', max_results=20)
ranked = rank_articles(items, query='artificial intelligence', method='auto')
clusters = cluster_articles(ranked, threshold=0.72, drop_singletons=True)
print('Clusters found:', len(clusters))
for cluster in clusters[:3]:
    print(cluster.get('label'), cluster.get('size'))

markdown = export.to_markdown(clusters, path='demo_output.md')
json_text = export.to_json(clusters, path='demo_output.json')
print('Export complete')
```
