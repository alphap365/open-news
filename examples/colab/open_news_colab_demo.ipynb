# Open News demo for Google Colab

This notebook installs the project, explores grouped fetch/search results, then ranks and exports story clusters.


```python
!pip install -q git+https://github.com/alphap365/open-news.git

from open_news import fetch, search

stories = fetch(category='tech', time_limit='w', max_results=5)
print('Grouped stories:', len(stories))
for story in stories[:3]:
    print('\n', story.get('title'), '->', story.get('url'))
    print('Sources:', ', '.join(story.get('sources', [])))
    for related in story.get('similar_articles', [])[:2]:
        print('  Related:', related.get('title'), '->', related.get('url'))

results = search('renewable energy', country='in', max_results=5)
print('\nSearch stories:', len(results))
for story in results[:3]:
    print(story.get('title'), '->', story.get('url'))
```

```python
from open_news import cluster_articles, export, fetch, rank_articles

items = fetch(category='tech', max_results=20)
ranked = rank_articles(items, query='artificial intelligence', method='auto')
clusters = cluster_articles(ranked, threshold=0.72, drop_singletons=True)
print('Clusters found:', len(clusters))
for cluster in clusters[:3]:
    print(cluster.get('label'), cluster.get('size'))

markdown = export.to_markdown(clusters, path='demo_output.md')
json_text = export.to_json(clusters, path='demo_output.json')
print('Export complete')
```
